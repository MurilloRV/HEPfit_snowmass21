```python
import numpy as np
import matplotlib.pyplot as plt
```

## Testing function to evalute $A_f$

```python
def A_f(f, sin2thetaEff):
    if f in ['u', 'c', 't']:
        T3 = 0.5
        Q = 2.0/3.0
    elif f in ['d', 's', 'b']:
        T3 = -0.5
        Q = -1.0/3.0
    elif f in ['e', 'mu', 'tau']:
        T3 = -0.5
        Q = -1.0
    else:
        raise ValueError("Invalid fermion type for A_f calculation")

    gV = T3 - 2*Q*sin2thetaEff
    gA = T3

    return 2*gV*gA/(gV**2 + gA**2)

def A_f_alt(f, sin2thetaEff):
    if f in ['u', 'c', 't']:
        Q = 2.0/3.0
    elif f in ['d', 's', 'b']:
        Q = -1.0/3.0
    elif f in ['e', 'mu', 'tau']:
        Q = -1.0
    else:
        raise ValueError("Invalid fermion type for A_f calculation")

    return (1 - 4*np.abs(Q)*sin2thetaEff) / (1 - 4*np.abs(Q)*sin2thetaEff + 8*np.abs(Q)**2*sin2thetaEff**2)
```

```python
sin2thetaEff = np.linspace(0.01, 0.99, 1000)
colors = {'u': 'blue', 'd': 'red', 'e': 'green'}
sin2thetaEff_SM = 0.2314833513035433

plt.figure(figsize=(6,5))
for f in ['u', 'd', 'e']:
    Af_values = A_f(f, sin2thetaEff)
    Af_alt_values = A_f_alt(f, sin2thetaEff)
    plt.plot(sin2thetaEff, Af_values, color=colors[f], label=fr"$A_{{ {f} }}$")
    plt.plot(sin2thetaEff, Af_alt_values, linestyle='dashed', color=colors[f])

plt.axvline(sin2thetaEff_SM, color='black', linestyle='dotted', label=r'$\sin^2\theta_{\mathrm{eff}}$ (Exp. value)')

plt.legend()
plt.xlabel(r'$\sin^2\theta_{\mathrm{eff}}$')
plt.ylabel(r'$A_f$')
```
